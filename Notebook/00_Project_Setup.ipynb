# 00 - Project Setup
تهيئة مسارات المشروع وملفات الـ calibration والمناطق. عدّل ملفات `Data/config` قبل تشغيل البيانات الحقيقية.

```python
from pathlib import Path
import json

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == 'Notebook':
    PROJECT_ROOT = PROJECT_ROOT.parent

for relative in ['Data/raw', 'Data/config', 'Notebook', 'Output/tables', 'Output/database', 'Output/alerts', 'Output/app', 'Output/videos']:
    (PROJECT_ROOT / relative).mkdir(parents=True, exist_ok=True)

config_dir = PROJECT_ROOT / 'Data' / 'config'
zones_path = config_dir / 'store_zones.json'
calibration_path = config_dir / 'camera_calibration.generated.json'
employee_ids_path = config_dir / 'employee_local_ids.json'

assert zones_path.exists(), 'Missing Data/config/store_zones.json'
assert employee_ids_path.exists(), 'Missing Data/config/employee_local_ids.json'

ZONES = json.loads(zones_path.read_text(encoding='utf-8'))
CALIBRATIONS = json.loads(calibration_path.read_text(encoding='utf-8')) if calibration_path.exists() else {}
print(f'Project root: {PROJECT_ROOT}')
print('Identity scope: camera_local (no cross-camera person matching)')
print('Configured zones:', [zone['zone_id'] for zone in ZONES])
print('Cameras with zone calibration:', list(CALIBRATIONS))
```
